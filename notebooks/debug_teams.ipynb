```python
import pandas as pd
from pathlib import Path

HIST_PATH = Path("../data_app/processed/all_leagues/all_leagues_features.csv")
hist = pd.read_csv(HIST_PATH)

hist_teams = (
    pd.Series(
        pd.concat([hist["HomeTeam"], hist["AwayTeam"]])
        .dropna()
        .unique()
    )
    .sort_values()
    .reset_index(drop=True)
)

print("Liczba drużyn w historii:", len(hist_teams))
hist_teams.head(30)
out_dir = Path("data_app/debug")
out_dir.mkdir(parents=True, exist_ok=True)

out_path = out_dir / "teams_from_history.csv"
hist_teams.to_csv(out_path, index=False, header=["team_name_history"], encoding="utf-8")

print(f" Zapisano nazwy historyczne: {out_path}")
print("Liczba drużyn historycznych:", len(hist_teams))
```
