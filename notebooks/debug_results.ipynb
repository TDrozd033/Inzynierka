```python
import pandas as pd

# === WCZYTANIE DANYCH ===
fixtures = pd.read_csv(
    "../data_app/prepared/premier_league_next_matchday_fixtures.csv",
    parse_dates=["Date"]
)

features = pd.read_csv(
    "../data_app/prepared/premier_league_next_matchday_features.csv"
)

# === POŁĄCZENIE ===
df = pd.concat([fixtures, features], axis=1)

print("Liczba meczów:", len(df))

# === MECZE Z BRAKAMI ===
df_nan = df[df.isna().any(axis=1)].copy()
print("Mecze z NaN:", len(df_nan))

# === PODSUMOWANIE (CZYTELNE) ===
summary = df_nan[[
    "Date", "HomeTeam", "AwayTeam", "Matchday", "League"
]].copy()

summary["NaN_count"] = df_nan.isna().sum(axis=1)

summary
```

```python
import pandas as pd
from pathlib import Path

# ścieżka do wygenerowanych cech
FEATURES_PATH = Path(
    "../data_app/prepared/premier_league_next_matchday_features.csv"
)

df = pd.read_csv(FEATURES_PATH)

print("=== INFO ===")
print("Shape:", df.shape)

print("\n=== LICZBA NaN W KOLUMNACH ===")
nan_cols = df.isna().sum().sort_values(ascending=False)
print(nan_cols[nan_cols > 0])

print("\n=== LICZBA NaN W WIERSZACH (MECZE) ===")
df_nan_rows = df.isna().sum(axis=1)
print(df_nan_rows)

print("\n=== PODSUMOWANIE ===")
print("Mecze z >=1 NaN:", (df_nan_rows > 0).sum(), "/", len(df))
print("Mecze bez NaN:", (df_nan_rows == 0).sum(), "/", len(df))
```
