```python
import pandas as pd
from pathlib import Path


X_PATH = Path("../data_app/prepared/premier_league_next_matchday_features.csv")
F_PATH = Path("../data_app/prepared/premier_league_next_matchday_fixtures.csv")

X = pd.read_csv(X_PATH)
fx = pd.read_csv(F_PATH)

print("X shape:", X.shape)
print("Fixtures shape:", fx.shape)


nan_pct = (X.isna().mean() * 100).sort_values(ascending=False)
print("\n=== TOP NaN % w kolumnach ===")
print(nan_pct.head(20))


row_nan = X.isna().sum(axis=1)
out = fx.copy()
out["NaN_count"] = row_nan

print("\n=== Mecze z największą liczbą NaN ===")
display(out.sort_values("NaN_count", ascending=False).head(10))


idxs = out.sort_values("NaN_count", ascending=False).head(3).index.tolist()
for i in idxs:
    missing_cols = X.columns[X.iloc[i].isna()].tolist()
    print(f"\n--- MATCH index={i} ---")
    print(out.loc[i, ["Date","HomeTeam","AwayTeam","Matchday","League","NaN_count"]])
    print("Missing cols:", missing_cols)


key_cols = [c for c in X.columns if any(k in c for k in [
    "Home_GF_last5","Home_GA_last5","Home_Pts_last5",
    "Away_GF_last5","Away_GA_last5","Away_Pts_last5",
    "SeasonStrength_PtsDiff","FormDiff_"
])]
print("\n=== Kluczowe kolumny snapshotów i ich NaN count ===")
print(X[key_cols].isna().sum().sort_values(ascending=False).head(30))


print("\n=== Podgląd wartości snapshotów (pierwsze 5 meczów) ===")
display(pd.concat([fx.head(5), X[key_cols].head(5)], axis=1))
```

```python
import pandas as pd
import numpy as np
from pathlib import Path


PROJECT_ROOT = Path.cwd().resolve().parents[0]  
print("PROJECT_ROOT:", PROJECT_ROOT)


HIST_PATH = PROJECT_ROOT / "data_app" / "processed" / "all_leagues" / "all_leagues_features.csv"
FEATURE_LIST_PATH = PROJECT_ROOT / "models" / "feature_list.json"


LEAGUE_KEY = "premier_league"
FUTURE_PATH = PROJECT_ROOT / "data_app" / "prepared" / f"{LEAGUE_KEY}_next_matchday_features.csv"
FIXTURES_PATH = PROJECT_ROOT / "data_app" / "prepared" / f"{LEAGUE_KEY}_next_matchday_fixtures.csv"

print("HIST_PATH exists:", HIST_PATH.exists(), HIST_PATH)
print("FUTURE_PATH exists:", FUTURE_PATH.exists(), FUTURE_PATH)
print("FIXTURES_PATH exists:", FIXTURES_PATH.exists(), FIXTURES_PATH)
print("FEATURE_LIST_PATH exists:", FEATURE_LIST_PATH.exists(), FEATURE_LIST_PATH)


hist = pd.read_csv(HIST_PATH, parse_dates=["Date"])
fixtures = pd.read_csv(FIXTURES_PATH, parse_dates=["Date"])
X_future = pd.read_csv(FUTURE_PATH)

feature_cols = pd.read_json(FEATURE_LIST_PATH, typ="series").tolist()


X_train = hist[feature_cols].copy()

missing_in_future = [c for c in feature_cols if c not in X_future.columns]
extra_in_future = [c for c in X_future.columns if c not in feature_cols]
print("\nMissing columns in X_future:", missing_in_future)
print("Extra columns in X_future:", extra_in_future)

X_future = X_future[feature_cols].copy()

def to_numeric_df(df: pd.DataFrame) -> pd.DataFrame:
    out = df.copy()
    for c in out.columns:
        out[c] = pd.to_numeric(out[c], errors="coerce")
    return out

X_train_num = to_numeric_df(X_train)
X_future_num = to_numeric_df(X_future)


def describe_compare(train_s: pd.Series, future_s: pd.Series) -> dict:
    return {
        "train_nonnull": int(train_s.notna().sum()),
        "future_nonnull": int(future_s.notna().sum()),
        "train_mean": float(train_s.mean(skipna=True)),
        "future_mean": float(future_s.mean(skipna=True)),
        "train_std": float(train_s.std(skipna=True)),
        "future_std": float(future_s.std(skipna=True)),
        "train_p05": float(train_s.quantile(0.05)),
        "future_p05": float(future_s.quantile(0.05)),
        "train_p50": float(train_s.quantile(0.50)),
        "future_p50": float(future_s.quantile(0.50)),
        "train_p95": float(train_s.quantile(0.95)),
        "future_p95": float(future_s.quantile(0.95)),
    }

rows = []
for c in feature_cols:
    rows.append({"feature": c, **describe_compare(X_train_num[c], X_future_num[c])})

cmp = pd.DataFrame(rows)


cmp["future_unique"] = [X_future_num[c].nunique(dropna=True) for c in feature_cols]
cmp["future_nan_pct"] = (X_future_num.isna().mean() * 100).values


cmp_sorted = cmp.sort_values(
    by=["future_nan_pct", "future_unique", "future_std"],
    ascending=[False, True, True]
)

print("\n=== TOP 15 podejrzanych cech (future: dużo NaN / mało unikalnych / małe std) ===")
display(cmp_sorted.head(15))

print("\n=== szybki podgląd X_future (num) ===")
display(X_future_num.head(10))


row_nan = X_future_num.isna().sum(axis=1)
print("\nNaN per match row:")
print(row_nan.value_counts().sort_index())

print("\nFixtures preview:")
display(fixtures[["Date","HomeTeam","AwayTeam","Matchday","League"]].head(10))
```

```python
import joblib

MODEL_PATH = PROJECT_ROOT / "models" / "decision_tree_model.pkl"
ENC_PATH = PROJECT_ROOT / "models" / "label_encoder.pkl"

model = joblib.load(MODEL_PATH)
le = joblib.load(ENC_PATH)

print("Model loaded:", MODEL_PATH.exists())
print("Encoder loaded:", ENC_PATH.exists())
print("Classes (encoder):", le.classes_)


proba = model.predict_proba(X_future_num)
pred_idx = np.argmax(proba, axis=1)
pred_cls = le.inverse_transform(pred_idx)

proba_df = pd.DataFrame(proba, columns=[f"P_{c}" for c in le.classes_])
out = pd.concat(
    [fixtures[["Date","HomeTeam","AwayTeam","Matchday","League"]].reset_index(drop=True),
     proba_df,
     pd.Series(pred_cls, name="Pred")],
    axis=1
)

display(out)



print("Unique rows of proba:", pd.DataFrame(proba).round(6).drop_duplicates().shape[0], "/", proba.shape[0])
print(proba_df.std().sort_values())
print(pd.Series(pred_cls).value_counts(normalize=True))
```
